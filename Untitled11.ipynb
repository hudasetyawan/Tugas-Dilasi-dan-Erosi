```python

# Step 1: Import semua library
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from PIL import Image

# Step 2: Upload gambar dari komputer
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Step 3: Baca gambar dan ubah ke grayscale
img = Image.open(filename).convert('L')
img_np = np.array(img)

# Step 4: Ubah ke biner (hitam putih)
_, binary = cv2.threshold(img_np, 127, 255, cv2.THRESH_BINARY_INV)

# Step 5: Buat daftar 9 structuring element
strels = [
    ("Rect 3x3", cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))),
    ("Ellipse 3x3", cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))),
    ("Cross 3x3", cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))),
    ("Rect 5x5", cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))),
    ("Ellipse 5x5", cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))),
    ("Cross 5x5", cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))),
    ("Horizontal 7x1", cv2.getStructuringElement(cv2.MORPH_RECT, (7, 1))),
    ("Vertical 1x7", cv2.getStructuringElement(cv2.MORPH_RECT, (1, 7))),
    ("Rect 7x7", cv2.getStructuringElement(cv2.MORPH_RECT, (7, 7))),
]

# Step 6: Simpan hasil ke list
results = [("Original", binary)]
for name, kernel in strels:
    dilated = cv2.dilate(binary, kernel)
    eroded = cv2.erode(binary, kernel)
    results.append((f"Erosi - {name}", dilated))
    results.append((f"Dilasi - {name}", eroded))

# Step 7: Tampilkan hasil dengan matplotlib
cols = 3
rows = int(np.ceil(len(results) / cols))
plt.figure(figsize=(16, rows * 4))
for i, (title, image) in enumerate(results):
    plt.subplot(rows, cols, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()
```
